# SIT320 - Task 3: Tic-Tac-Toe Playing Algorithm

The human plays **X**, the computer plays **O**.

**Strategy:** the computer uses the **minimax** algorithm. Before every move it
recursively explores all possible future game states, assuming both players play
perfectly, and picks the move with the best guaranteed outcome. Scores are
depth-adjusted (`10 - depth` for a win, `depth - 10` for a loss) so the computer
prefers fast wins and slow losses. Because Tic-Tac-Toe has at most 9! ≈ 362,880
game sequences, a full search is cheap and the computer never loses — the best a
human can achieve against it is a draw.

See `task3_pseudocode.md` for the full pseudocode and complexity analysis.

## Board setup and helpers

```python
HUMAN = "X"
COMPUTER = "O"
LINES = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
    (0, 4, 8), (2, 4, 6),             # diagonals
]


def print_board(board):
    for row in range(3):
        cells = [board[3 * row + col] or str(3 * row + col + 1) for col in range(3)]
        print(" " + " | ".join(cells))
        if row < 2:
            print("---+---+---")
    print()


def winner(board):
    for a, b, c in LINES:
        if board[a] and board[a] == board[b] == board[c]:
            return board[a]
    return None
```

## Minimax search

```python
def minimax(board, depth, is_maximising):
    won = winner(board)
    if won == COMPUTER:
        return 10 - depth
    if won == HUMAN:
        return depth - 10
    if all(board):
        return 0

    if is_maximising:  # computer's turn: take the best score for itself
        best = -100
        for i in range(9):
            if not board[i]:
                board[i] = COMPUTER
                best = max(best, minimax(board, depth + 1, False))
                board[i] = None
        return best
    else:  # human's turn: assume they play their best reply
        best = 100
        for i in range(9):
            if not board[i]:
                board[i] = HUMAN
                best = min(best, minimax(board, depth + 1, True))
                board[i] = None
        return best


def best_move(board):
    best_score, move = -100, None
    for i in range(9):
        if not board[i]:
            board[i] = COMPUTER
            score = minimax(board, 0, False)
            board[i] = None
            if score > best_score:
                best_score, move = score, i
    return move
```

## Game loop

```python
def play():
    board = [None] * 9
    print("Tic-Tac-Toe: you are X, the computer is O. Pick a cell 1-9.\n")
    print_board(board)

    current = HUMAN
    while winner(board) is None and not all(board):
        if current == HUMAN:
            choice = input("Your move (1-9): ").strip()
            if not choice.isdigit() or not 1 <= int(choice) <= 9 or board[int(choice) - 1]:
                print("Invalid move, try again.")
                continue
            board[int(choice) - 1] = HUMAN
        else:
            move = best_move(board)
            board[move] = COMPUTER
            print(f"Computer plays cell {move + 1}:")
        print_board(board)
        current = COMPUTER if current == HUMAN else HUMAN

    won = winner(board)
    print(f"{won} wins!" if won else "It's a draw!")
```
