# SIT320 - Task 2c: Basic Python Script

Demonstrates `while` and `if` statements: counts down from a starting number,
classifying each value along the way, and keeps a running total of the even numbers seen.

Running in the **SIT320** conda environment.

```python
n = 10
even_total = 0

while n > 0:
    if n % 2 == 0:
        even_total += n
        print(f"{n} is even (running even total: {even_total})")
    elif n % 3 == 0:
        print(f"{n} is odd and divisible by 3")
    else:
        print(f"{n} is odd")
    n -= 1

print(f"Countdown finished. Sum of even numbers 1-10: {even_total}")
```
