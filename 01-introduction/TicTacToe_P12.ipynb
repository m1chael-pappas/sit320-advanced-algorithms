## Board Definition

```python
#define the board
board={1:' ', 2:' ', 3:' ',
       4:' ', 5:' ', 6:' ',
       7:' ', 8:' ', 9:' '}
```

```python
#function to print the board
def printBoard(board):
    print(board[1]+'|'+board[2]+'|'+board[3])
    print('-+-+-')
    print(board[4] + '|' + board[5] + '|' + board[6])
    print('-+-+-')
    print(board[7] + '|' + board[8] + '|' + board[9])
    print('\n')
```

```python
printBoard(board)
```

```python
#function to check is a certain position in the board is empty.
def spaceIsFree(position):
    
    if (board[position]==' '):
        return True
    else:
        return False
```

```python
#method to insert letter in space
def insertLetter(letter, position):
    
    if (spaceIsFree(position)):
        board[position]=letter
        printBoard(board)
        
        #I check for a win before a draw now. If the winning move fills the
        #last empty cell, chkDraw() is also true at that point, so checking
        #the draw first was calling those games a draw.
        if (chkForWin()):
            if (letter=='X'):
                print('Bot wins!')
            else:
                print('You win!')
        elif (chkDraw()):
            print('Draw!')
        return
    
    else:
        print('Position taken, please pick a different position.')
        position=int(input('Enter new position: '))
        insertLetter(letter, position)
        
        return
```

```python
#function to check if board is draw
def chkDraw():
    
    for key in board.keys():
        if (board[key]==' '):
            return False
    return True

#function to check if one user has won
def chkForWin():
    
    if (board[1]==board[2] and board[1]==board[3] and board[1] !=' '):
        return True
    elif (board[4]==board[5] and board[4]==board[6] and board[4] !=' '):
        return True
    elif (board[7]==board[8] and board[7]==board[9] and board[7] !=' '):
        return True
    elif (board[1]==board[5] and board[1]==board[9] and board[1] !=' '):
        return True
    elif (board[3]==board[5] and board[3]==board[7] and board[3] !=' '):
        return True
    elif (board[1]==board[4] and board[1]==board[7] and board[1] !=' '):
        return True
    elif (board[2]==board[5] and board[2]==board[8] and board[2] !=' '):
        return True
    elif (board[3]==board[6] and board[3]==board[9] and board[3] !=' '):
        return True
    else:
        return False
```

```python
#function to check who won
def chkMarkForWin(mark):
    
    if (board[1]==board[2] and board[1]==board[3] and board[1] ==mark):
        return True
    elif (board[4]==board[5] and board[4]==board[6] and board[4] ==mark):
        return True
    elif (board[7]==board[8] and board[7]==board[9] and board[7] ==mark):
        return True
    elif (board[1]==board[5] and board[1]==board[9] and board[1] ==mark):
        return True
    elif (board[3]==board[5] and board[3]==board[7] and board[3] ==mark):
        return True
    elif (board[1]==board[4] and board[1]==board[7] and board[1] ==mark):
        return True
    elif (board[2]==board[5] and board[2]==board[8] and board[2] ==mark):
        return True
    elif (board[3]==board[6] and board[3]==board[9] and board[3] ==mark):
        return True
    else:
        return False
```

```python
player = 'O'
bot ='X'
```

```python
#function for player move
def playerMove():
    position=int(input('Enter position for O:'))
    insertLetter(player, position)
    return

#function for bot move
def compMove():

    #Task 5: before running minimax I scan the board for a move that wins
    #on the spot. If one exists the bot just plays it instead of searching
    #the whole game tree.
    for key in board.keys():
        if (board[key]==' '):
            board[key]=bot
            if (chkMarkForWin(bot)):
                board[key]=' '
                print('Bot found a winning move at position', key,
                      'and skipped the minimax search.')
                insertLetter(bot, key)
                return
            board[key]=' '

    bestScore=-1000 #lowest to start with, for comaparison to the score after a certain move

    bestMove=0

    for key in board.keys(): #for loop to find emplty slots, make a move, calculate the score and see if its higher then best score
        if (board[key]==' '):
            board[key]=bot
            score = minimax(board, 1, False) #calculate the score
            board[key] = ' ' #set board back to what it was
            if (score > bestScore): #compare the score from a certain move to the best score.
                bestScore = score
                bestMove = key

    #at the end of the loop, we have the best move figured out.
    insertLetter(bot, bestMove)
    return
```

```python
def minimax(board, depth, isMaximizing):

    #Task 5: I adjust the scores by depth so a win in fewer moves is worth
    #more. This way the search itself also leans towards quick wins and
    #drags out losses as long as possible.
    if (chkMarkForWin(bot)):
        return 10 - depth
    elif (chkMarkForWin(player)):
        return depth - 10
    elif (chkDraw()):
        return 0

    if isMaximizing:

        bestScore = -1000

        for key in board.keys():
            if board[key]==' ':
                board[key]=bot
                score = minimax(board, depth+1, False) #depth was stuck at 0 before, now it actually counts up
                board[key]= ' '
                if (score>bestScore):
                    bestScore = score
        return bestScore

    else:

        bestScore = 1000

        for key in board.keys():
            if board[key]==' ':
                board[key]=player
                score = minimax(board, depth+1, True) #depth was stuck at 0 before, now it actually counts up
                board[key]= ' '
                if (score<bestScore):
                    bestScore = score
        return bestScore
```

```python
#the loop now stops on a draw as well as a win, and I check the game state
#again after the bot moves so it doesn't ask me for a move when the game
#is already over
while not chkForWin() and not chkDraw():
    compMove()
    if chkForWin() or chkDraw():
        break
    playerMove()
```

## Test scenario: the bot takes an instant win instead of searching

To show the Task 5 change working, I wrote a quick test scenario. I set up a board where the bot (X) already has two in a row and can win straight away at position 3. I also wrapped `minimax()` with a small call counter so I can see whether the search runs. For this move the counter stays at 0, so the bot never touched minimax and just played the winning cell.

I then ran a second board with no instant win available. There the counter jumps into the thousands because the full minimax search has to run.

```python
#quick call counter around minimax so I can see when the search actually runs
if 'minimax_original' not in globals():
    minimax_original = minimax

minimaxCalls = 0
def minimax(board, depth, isMaximizing):
    global minimaxCalls
    minimaxCalls += 1
    return minimax_original(board, depth, isMaximizing)

#scenario 1: the bot (X) can win straight away at position 3
board = {1:'X', 2:'X', 3:' ',
         4:'O', 5:'O', 6:' ',
         7:' ', 8:' ', 9:' '}
print('Scenario 1, board before the bot moves (X = bot, O = player):')
printBoard(board)
compMove()
print('minimax() calls for this move:', minimaxCalls)

#scenario 2: no instant win on the board, so the full search has to run
minimaxCalls = 0
board = {1:'X', 2:' ', 3:' ',
         4:' ', 5:'O', 6:' ',
         7:' ', 8:' ', 9:' '}
print('\nScenario 2, board with no instant win available:')
printBoard(board)
compMove()
print('minimax() calls for this move:', minimaxCalls)

minimax = minimax_original #put the normal minimax back
```
